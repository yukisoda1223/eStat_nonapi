```python
import os
import requests

# ===== 設定 =====
base = 32163095          # 000032163095 の下8桁
count = 47               # 何本取るか（n=1..count）
STEP = 15                # +15ずつ
OUT_DIR = r"C:...\Downloads\zenkoku\人口データ\Raw"
# =================

os.makedirs(OUT_DIR, exist_ok=True)
session = requests.Session()

saved = 0
skipped_404 = 0
skipped_non_csv = 0
errors = 0

for i in range(count):
    n = i + 1  # 1始まり（01..47）
    id8 = base + STEP * i
    statInfId = f"0000{id8:08d}"
    url = f"https://www.e-stat.go.jp/stat-search/file-download?statInfId={statInfId}&fileKind=1"

    # 保存名：h03_01.csv ... h03_47.csv（nを2桁ゼロ埋め）
    outname = f"h03_{n:02d}.csv"
    outpath = os.path.join(OUT_DIR, outname)

    try:
        r = session.get(url, allow_redirects=True, timeout=60)

        if r.status_code == 404:
            skipped_404 += 1
            print(f"404 skip: {outname} statInfId={statInfId} url={url}")
            continue

        r.raise_for_status()

        # 簡易チェック：HTMLが返ってきたらスキップ（ログイン誘導等）
        head = r.content[:200].lstrip()
        if head.startswith(b"<"):
            skipped_non_csv += 1
            print(f"non-csv? skip: {outname} statInfId={statInfId} (looks like HTML)")
            continue

        with open(outpath, "wb") as f:
            f.write(r.content)

        saved += 1
        print(f"saved: {outname} (statInfId={statInfId}) -> {outpath}")

    except requests.RequestException as e:
        errors += 1
        print(f"error: {outname} statInfId={statInfId} {e}")

print("\n===== summary =====")
print(f"target count         : {count}")
print(f"saved                : {saved}")
print(f"skipped (404)        : {skipped_404}")
print(f"skipped (non-csv)    : {skipped_non_csv}")
print(f"errors               : {errors}")
print(f"output dir           : {OUT_DIR}")
```
